## Process demographic data

```python
demogr_df = demogr_df.loc[demogr_df["SEQN_new"].str.contains("C")]

demogr_cols = demogr_df.columns.values.tolist()
demogr_cols = [col for col in demogr_cols if "WTMREP" not in col]
demogr_cols = [col for col in demogr_cols if "WTIREP" not in col]
demogr_cols = demogr_cols[2:]

print(len(demogr_cols))
# print(demogr_cols)
```

```python
demogr_features_df = demogr_df.loc[:, ["SEQN", "SEQN_new", "DMDCITZN", "DMDEDUC2", "DMDFMSIZ", "DMDMARTL", "INDFMPIR", "RIAGENDR", "RIDAGEYR"]]
print(demogr_features_df.shape)
demogr_features_df.head()
```

## Process questionnaire data (stroke data)

```python
stroke_df = qs_df.loc[:, ["SEQN", "SEQN_new", "MCD180F"]]
# stroke_df = stroke_df.dropna()
stroke_df["SEQN"] = stroke_df["SEQN"].astype(int)
# stroke_df["MCD180F"] = stroke_df["MCD180F"].astype(int)

print(stroke_df.shape)
stroke_df.head()
```

## Make feature_df

```python
# feature_df = pd.concat([demogr_features_df, stroke_df], keys=["SEQN", "SEQN_new"])
feature_df = pd.merge(demogr_features_df, stroke_df, on=["SEQN", "SEQN_new"])
print(feature_df.shape)
feature_df.head()
```

```python
feature_df.describe()
```

```python
feature_df = feature_df.drop(feature_df[feature_df["MCD180F"] == 99999].index)
print(feature_df.shape)
feature_df.head()
```

```python
feature_df = feature_df.dropna(subset=["SEQN", "SEQN_new", "DMDCITZN", "DMDEDUC2", "DMDFMSIZ", "DMDMARTL", "INDFMPIR", "RIAGENDR", "RIDAGEYR"])
print(feature_df.shape)
feature_df.head()
```

```python
feature_df = feature_df[(feature_df["SEQN_new"].str.contains("C")) & (feature_df["RIDAGEYR"] >= 55)]
feature_df["stroke"] = feature_df["MCD180F"].notna().astype(int)
print(feature_df.shape)
feature_df.head()
```

```python
feature_df.describe()
```

```python
feature_df = feature_df.drop(["SEQN", "MCD180F"], axis=1)
print(feature_df.shape)
feature_df.head()
```

```python
feature_df = feature_df.rename(columns={
    "DMDCITZN": "US_citizenship",
    "DMDEDUC2": "education_level",
    "DMDFMSIZ": "num_familiy",
    "DMDMARTL": "marital_status",
    "INDFMPIR": "PIR",
    "RIAGENDR": "gender",
    "RIDAGEYR": "age"
})
feature_df.head()
```

```python
# feature_df = feature_df.drop(feature_df[feature_df["US_citizenship"] == 9].index)
# feature_df = feature_df.drop(feature_df[feature_df["US_citizenship"] == 7].index)

feature_df["US_citizenship"] = feature_df["US_citizenship"].apply(lambda x: np.nan if x == 9 or x == 7 else x)
feature_df = feature_df.replace({"US_citizenship": {2.0: 0.0}})
# feature_df["US_citizenship"] = feature_df["US_citizenship"].astype(int)
print(feature_df.shape)
feature_df.head()
```

```python
feature_df[feature_df["US_citizenship"].isnull()]
```

```python
# feature_df = feature_df.drop(feature_df[feature_df["education_level"] == 9].index)
# feature_df = feature_df.drop(feature_df[feature_df["education_level"] == 7].index)
# feature_df["education_level"] = feature_df["education_level"].astype(int)
feature_df["education_level"] = feature_df["education_level"].apply(lambda x: np.nan if x == 9 or x == 7 else x)
print(feature_df.shape)
feature_df.head()
```

```python
feature_df[feature_df["education_level"].isnull()]
```

```python
feature_df["num_familiy"] = feature_df["num_familiy"].astype(int)
print(feature_df.shape)
feature_df.head()
```

```python
# feature_df = feature_df.drop(feature_df[feature_df["marital_status"] == 99].index)
# feature_df = feature_df.drop(feature_df[feature_df["marital_status"] == 77].index)
# feature_df["marital_status"] = feature_df["marital_status"].astype(int)
feature_df["marital_status"] = feature_df["marital_status"].apply(lambda x: np.nan if x == 99 or x == 77 else x)
print(feature_df.shape)
feature_df.head()
```

```python
feature_df[feature_df["marital_status"].isnull()]
```

```python
feature_df = feature_df.replace({"gender": {1: 0, 2: 1}})
print(feature_df.shape)
feature_df.head()
```
